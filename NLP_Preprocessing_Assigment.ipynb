```python
import string
string.punctuation
```

```python
data['text']="".join([i for i in data['text'] if i  not in string.punctuation])
```

```python
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords', quiet=True)

stop_words = set(stopwords.words('english'))

def remove_stopwords(text):
    if not isinstance(text, str):
        return ""

    return " ".join(
        word for word in text.split()
        if word.lower() not in stop_words
    )

# Test on only 100 rows
data.loc[:99, 'clean_text'] = (
    data.loc[:99, 'text'].apply(remove_stopwords)
)

data[['text', 'clean_text']].head()
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer

# Get text from the entire dataset
texts = data['clean_text'].fillna('').astype(str).tolist()

# Create tokenizer
tokenizer = Tokenizer()

# Learn vocabulary from all text
tokenizer.fit_on_texts(texts)

# Convert all text into numerical sequences
sequences = tokenizer.texts_to_sequences(texts)

# Display results
print("Total texts:", len(texts))
print("Vocabulary Size:", len(tokenizer.word_index))
print("First 5 words:", list(tokenizer.word_index.items())[:5])
print("First 5 sequences:", sequences[:5])
```
